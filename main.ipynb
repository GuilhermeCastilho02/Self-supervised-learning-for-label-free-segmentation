Definição das funções que geram as weak labels e que calculam o QC (Quality Control).

```python
def generate_weak_mask(img_path):
    # Carrega a imagem e redimenciona
    img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
    h, w = img.shape
    img = cv2.resize(img, (512, 512))
    # "Estoura" pixels extremos para que não entrem na máscara
    img[img < 50] = 255 
    img[img > 180] = 255
    
    # Equalização + suavização
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    I = clahe.apply(img)
    I = cv2.GaussianBlur(I, (7,7), 0)
    
    for ang in np.arange(0, 151, 30):
        se = np.zeros((15, 15), np.uint8)
        cv2.ellipse(se, (15//2, 15//2), (15//2, 1), ang, 0, 360, 1, -1)
        I_supp = cv2.morphologyEx(I, cv2.MORPH_OPEN, se)
        I = cv2.min(I, I_supp)
        
    # Inverte para que opacidades fiquem claras
    I_inv = cv2.normalize(255 - I, None, 0, 255, cv2.NORM_MINMAX)

    # Filtro bilateral
    I_inv_bilateral = cv2.bilateralFilter(I_inv.astype(np.uint8), d=-1, sigmaColor=0.25*255, sigmaSpace=15)

    # Aplica um treshold simples
    _, bw = cv2.threshold(I_inv_bilateral, 127, 255, cv2.THRESH_BINARY)
    
    # Limpa regiões fora dos pulmões
    # remove um pequeno pedaço do topo e das laterais que não incluem o pulmão (geralmente)
    bw[:40, :], bw[:, -70:], bw[:, :70] = 0,0, 0
    
    # Morfologia para suavizar
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9,9))
    bw = cv2.morphologyEx(bw, cv2.MORPH_OPEN, kernel)
    bw = cv2.morphologyEx(bw, cv2.MORPH_CLOSE, kernel)
    
    # Mantém regiões internas
    num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(bw)
    mask_out = np.zeros_like(bw)
    
    for i in range(1, num_labels):
        area = stats[i, cv2.CC_STAT_AREA]
        x,y = stats[i, cv2.CC_STAT_LEFT], stats[i, cv2.CC_STAT_TOP]
        wc,hc = stats[i, cv2.CC_STAT_WIDTH], stats[i, cv2.CC_STAT_HEIGHT]

        # Descarta regiões no topo e muito pequenas
        if area > 500 and y > 39:
            mask_out[labels == i] = 255

    # Remove objetos pequenos (geralmente ombro ou parte da silhueta lateral)
    mask_out = morphology.remove_small_objects(mask_out.astype(bool), min_size=5000)
    
    # Volta ao tamanho original
    mask_out = mask_out.astype(np.uint8)
    mask_out = cv2.resize(mask_out, (w, h))
    
    return torch.from_numpy(mask_out).unsqueeze(0).float()

# Função QC
def qc_score(mask, min_area=0.08, max_area=0.18, min_ecc=0.85, max_ecc=0.97, 
             min_solidity=0.8, max_solidity=0.97, min_aspect=0.35, max_aspect=0.65, num_obj=2):
    props = measure.regionprops(measure.label(mask[0]))
    # Apenas os 2 maiores objetos (idealmente apenas pulmões)
    props = sorted(props, key=lambda x: x.area, reverse=True)[:2]

    H, W = mask[0].shape

    scores = 0

    try:
        biggest_area = props[0].area
    except:
        # print(f"Mascara problemática, Nº de objetos nela: {len(props)}")
        return 0

    for p in props:
        lung_scores = list()
        # Área relativa
        area_frac = p.area / (H * W)
        if area_frac < min_area * 0.7:
            lung_scores.append(0)
        elif area_frac < min_area:
            lung_scores.append(area_frac / min_area / 2)
        elif area_frac > max_area * 1.4:
            lung_scores.append(0)
        elif area_frac > max_area:
            lung_scores.append(max_area / area_frac / 2)
        else:
            lung_scores.append(1)

        # Eccentricidade
        ecc = p.eccentricity
        if ecc < min_ecc * 0.8:
            lung_scores.append(0)
        elif ecc < min_ecc:
            lung_scores.append(ecc)
        elif ecc > max_ecc:
            lung_scores.append(max_ecc - 0.15)
        else:
            lung_scores.append(1.0)

        # Solidity
        solidity=p.solidity
        if solidity < min_solidity * 0.75:
            lung_scores.append(0)
        elif solidity < min_solidity - 0.05:
            lung_scores.append(solidity * 0.75)
        elif solidity < min_solidity:
            lung_scores.append(solidity)
        elif solidity > max_solidity:
            lung_scores.append(solidity - 0.1)
        else:
            lung_scores.append(1.0)
            
        # # Extent
        # lung_scores.append(p.extent)

        # Aspect ratio
        minr, minc, maxr, maxc = p.bbox
        bbox_h = maxr - minr
        bbox_w = maxc - minc
        aspect = max(bbox_h / (bbox_w + 1e-9), bbox_w / (bbox_h + 1e-9))
        if aspect < min_aspect * 0.8:
            lung_scores.append(0)
        elif aspect < min_aspect:
            lung_scores.append(aspect / min_aspect * 0.75)
        elif aspect > 0.94:
            lung_scores.append(0)
        elif aspect > 0.82:
            lung_scores.append(max_aspect / aspect * 0.8)
        elif aspect > max_aspect:
            lung_scores.append(max_aspect / aspect * 0.9)
        else:
            lung_scores.append(1)

        # Nº de objetos
        n_obj = len(props)
        if n_obj > num_obj:
            lung_scores.append(1 / n_obj)
        else:
            lung_scores.append(n_obj / num_obj)
       
        # Soma o score do pulmão ao score final ponderado pela diferença de tamanho entre pulmões 
        scores += np.mean(lung_scores) * (p.area / biggest_area)

    return scores / 2 # Média dos pulmões
```

Classe de Dataset e Augmentation

```python
class SelfSegCXRDataset(Dataset): # TODO: Estou passando augmentation no teste e validação, AJUSTAR ISSO
    def __init__(self, dataset_path=INPUT_DIR, image_size=None, augment=None, device=DEVICE):
        self._path_data = dataset_path
        self.images_paths = sorted(glob(os.path.join(self._path_data, "images", "*.jpg")))#[:20000]
        self.weak_labels_paths = sorted(glob(os.path.join(self._path_data, "weak_labels", "*.png")))#[:20000]

        self.device = device
        self.augment = augment

        if image_size:
            self.resize_image = v2.Resize(size=image_size, interpolation=v2.InterpolationMode.BILINEAR, antialias=True)
            self.resize_mask = v2.Resize(size=image_size, interpolation=v2.InterpolationMode.NEAREST)
        else:
            self.resize_image = lambda x: x
            self.resize_mask = lambda x: x

        self.qcs = list()
        self.__generate_qcs()
        # ## NOVO TRECHO PROVISÓRIO
        # self.__data = list()

        # for index in tqdm(range(len(self.images_paths)), "Calculando QC Scores"):
        #     image_path = self.images_paths[index]
        #     mask_path = self.weak_labels_paths[index]
    
        #     image = decode_image(image_path).to(torch.float)
        #     mask = decode_image(mask_path).to(torch.float) / 255.0
            
        #     image = self.resize_image(image)
        #     mask = self.resize_mask(mask)
            
        #     qc = qc_score(mask)
    
        #     self.__data.append((image.to(torch.uint8), mask, qc))
            
        
    def __len__(self):
        return len(self.images_paths)

    def __generate_qcs(self):
        for index in tqdm(range(len(self.images_paths))):
            mask_path = self.weak_labels_paths[index]
            mask = decode_image(mask_path)
            mask = self.resize_mask(mask)
            
            qc = qc_score(mask)
            self.qcs.append(qc)

    def __getitem__(self, index):
        image_path = self.images_paths[index]
        mask_path = self.weak_labels_paths[index]

        image = decode_image(image_path)
        mask = (decode_image(mask_path) > 0).to(torch.uint8)
        
        image = self.resize_image(image)
        mask = self.resize_mask(mask)
        
        # qc = qc_score(mask)
        qc = self.qcs[index]
        
        # image, mask, qc = self.__data[index] #Trecho novo

        if self.augment:
            image, mask = self.augment(image, mask)
        
        # return image, mask, qc
        return image, mask, qc


class CXRAugmentation: # TODO: POLIR AUGMENTATION no sentido de sentido das probs ( self._translade_prob + 1 > 1 sempre pô)
    def __init__(self, flip_prob=0.5, rotate_prob=0.5, translade_prob=0.5, change_quality_prob=0.3):
        self._flip_prob = flip_prob
        self._rotate_prob = rotate_prob
        self._apply_prob = change_quality_prob
        self._translade_prob = translade_prob

    def apply(self, image, mask):
        augment_list = list()

        if np.random.rand() < self._flip_prob:
            augment_list.append(lambda img, msk: (TF.hflip(img), TF.hflip(msk)))

        if np.random.rand() < self._translade_prob + 1: # 20/11 tranlações de até 15% 23/11 testando com valor antigo
            prop_x = float(np.random.choice(np.arange(-5, 5.5, 0.5))) / 100
            prop_y = float(np.random.choice(np.arange(-5, 5.5, 0.5))) / 100

            H, W = image[0].shape
            translate_props = (int(prop_x * W), int(prop_y * H))

            augment_list.append(lambda img, msk: (TF.affine(img, 0, translate_props, 1, (0, 0)), TF.affine(msk, 0, translate_props, 1, (0, 0))))

        if np.random.rand() < self._rotate_prob + 1: # Tá rotacionando sempre (+ 1, igual na translação); até 15° 20/11
            angle = int(np.random.choice(np.arange(-5, 5)))
            augment_list.append(lambda img, msk: (TF.rotate(img, angle), TF.rotate(msk, angle)))


        if len(augment_list) < 1:
            augment_list.append(lambda x, y: (x, y))

        image, mask = v2.Compose(augment_list)(image, mask)

        image = v2.RandomApply([
            v2.ColorJitter(brightness=0.1, contrast=0.1),
            v2.GaussianBlur(kernel_size=3)
        ], p=self._apply_prob)(image)

        return image, mask

    def __call__(self, image, mask):
        return self.apply(image, mask)
```

Função para fazer o split do dataset

```python
class AugmentedSubset(Dataset):
    def __init__(self, subset, augment=None):
        self.subset = subset
        self.augment = augment

    def __len__(self):
        return len(self.subset)

    def __getitem__(self, idx):
        image, mask, qc = self.subset[idx]
        if self.augment is not None:
            image, mask = self.augment(image, mask)
        return image, mask, qc

def get_train_test_split(dataset, test_prop=0.2, seed=None):
    n_features = len(dataset)

    test_size = round(n_features * test_prop)
    train_size = n_features - test_size

    generator = torch.Generator().manual_seed(seed) if seed is not None else None
    
    train_raw, test_ds = random_split(dataset, [train_size, test_size], generator=generator)
    train_ds = AugmentedSubset(train_raw, augment=CXRAugmentation())

    return train_raw, train_ds, test_ds #23/11
```

## UNet

Loss function (Apesar de implementada a Soft Dice Loss que utiliza os QC Scores como pesos, optamos por utilizar uma Dice Loss simples já implementada)

```python
class WeightedSoftDiceLoss(nn.Module):
    def _init_(self, smooth=1e-6):
        super()._init_()
        self.smooth = smooth

    def forward(self, preds, targets, weights=None):
        """
        preds:   [B, 1, H, W]
        targets: [B, 1, H, W]
        weights: [B]
        """

        preds = preds.reshape(preds.size(0), -1)      # [B, HW]
        targets = targets.reshape(targets.size(0), -1)

        intersection = (preds * targets).sum(dim=1)    # [B]
        sums = preds.sum(dim=1) + targets.sum(dim=1)   # [B]

        dice = (2 * intersection + self.smooth) / (sums + self.smooth)  # [B]
        loss = 1 - dice                                                   # [B]

        if weights is not None:
            # normaliza pesos para manter a loss estável
            weights = weights / (weights.sum() + self.smooth)
            loss = loss * weights
            return loss.sum()

        return loss.mean()
```

Função de treino

```python
def train_unet(model, train_loader, loss, optimizer, num_epochs=5, use_weights=False, device=DEVICE, save_path=None):
    
    model = model.to(device)

    best_loss = float("inf")
    best_state = None

    history_loss = list()

    for epoch in range(num_epochs):
        model.train()
        total = 0

        # print(f"\nIniciando época {epoch + 1}/{num_epochs}")

        # tqdm
        progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}")
        
        for imgs, masks, qc in progress_bar:
            imgs = imgs.to(device, dtype=torch.float32) / 255
            masks = masks.to(device, dtype=torch.float32)

            # QC opcional como pesos
            weights = qc.to(device).float() if use_weights else None

            logits = model(imgs)
            loss_value = loss(logits, masks)
            # loss_value = loss(preds, masks, weights=weights)

            optimizer.zero_grad()
            loss_value.backward()
            optimizer.step()

            total += loss_value.item()

            # Atualiza barra com a loss atual
            progress_bar.set_postfix(loss=loss_value.item())

        epoch_loss = total / len(train_loader)
        print(f"Loss médio da época {epoch + 1}: {epoch_loss:.4f}")
        history_loss.append(epoch_loss)
        # Melhor chackpoint
        if epoch_loss < best_loss:
            best_loss = epoch_loss
            best_state = model.state_dict()
            # print(f"Novo melhor modelo! Loss = {best_loss:.4f}")

    # Salva modelo
    if save_path is not None and best_state is not None:
        torch.save(best_state, save_path)
        print(f"Modelo salvo  em: {save_path}")

    return model, history_loss
```

Filtro QC

```python
def qc_filter(dataset, threshold):
    filtered = list()
    for img, mask, qc in tqdm(dataset, desc="Filtro QC"):
        if qc > threshold:
            filtered.append((img, mask.to(torch.uint8), qc))
    return filtered
```

## Pipeline

Definições iniciais do modelo

```python
# Early Learning com as weak masks

threshold = 0.6
MAX_SELF_LEARNING_CYCLE = 5
MIN_DATA = len(DATASET) // 1.5
BATCH_SIZE = 16
NUM_WORKERS = 2

model = smp.Unet(encoder_name="resnet34", encoder_weights="imagenet", in_channels=1, classes=1, activation=None) # Mudança 23/11

loss = smp.losses.DiceLoss(mode="binary")#WeightedSoftDiceLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

train_raw, train_dataset, val_dataset = get_train_test_split(DATASET)
train_eval_loader = DataLoader(train_raw, batch_size=BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=True, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=True, shuffle=True)

model = model.to(DEVICE)
```

Early Learning e Self-learning

```python
train_filtered = qc_filter(train_raw, threshold)
train_filtered_aug = AugmentedSubset(train_filtered, augment=CXRAugmentation())
train_loader_loop = DataLoader(train_filtered, batch_size=BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=True, shuffle=True)

# Early Learning
model, loss_1 = train_unet(model, train_loader_loop, loss, optimizer, num_epochs=5, use_weights=False)

losses = dict()

losses["Loss 1"] = loss_1

for i in range(MAX_SELF_LEARNING_CYCLE): # Self-Learning
    print(f"iteração nº{i+1}")
    model.eval()
    new_train_data = list()

    with torch.no_grad():
        for imgs, masks, qcs in tqdm(train_eval_loader, desc="Previsões"):
            imgs_gpu = imgs.to(DEVICE, dtype=torch.float32) / 255#.unsqueeze(0)

            logits = model(imgs_gpu)
            probs = torch.sigmoid(logits)
            pred_masks = (probs > 0.5).to(torch.uint8).cpu()
            
            pred_masks = pred_masks
            # Passando o QC Score na CPU a cada iteração
            for j in range(imgs.size(0)):
                img_orig = imgs[j]
                pred_mask = pred_masks[j]
                qc_pred = qc_score(pred_mask)
                
                if qc_pred > threshold:
                    new_train_data.append((img_orig, pred_mask, qc_pred))
    
    print(f"Quantidade de mascaras aprovadas pelo QC: {len(new_train_data)}")

    new_train_aug = AugmentedSubset(new_train_data, augment=CXRAugmentation())
    train_loader_loop = DataLoader(new_train_aug, batch_size=BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=True, shuffle=True)
    
    # Re-treina a rede
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)
    model, loss_sl = train_unet(model, train_loader_loop, loss, optimizer, num_epochs=5)

    # Avaliação no conjunto de validação
    val_loss = 0
    model.eval()
    with torch.no_grad():
        for imgs, masks, qc in tqdm(val_loader, desc="Validação"):
            imgs = imgs.to(DEVICE, dtype=torch.float32) / 255 
            masks = masks.to(DEVICE, dtype=torch.float32)
            preds = model(imgs)
            val_loss += loss(preds, masks).item()

    avg_val_loss = val_loss / len(val_loader)
    print(f"Loss médio na validação: {avg_val_loss:.4f}")
    
    if len(new_train_data) > MIN_DATA:
        print("Parando: atingiu MIN_DATA")
        break

torch.save(model.state_dict(), "self_learning_unet.pth")

losses["Loss SL"] = loss_sl
losses["Loss Val SL"] = val_loss
```
